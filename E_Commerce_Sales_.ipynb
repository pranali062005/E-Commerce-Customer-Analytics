# **Feature Engineering**

```python
df['Birth_Year'] = df['Date_of_Birth'].dt.year
df[['Date_of_Birth','Birth_Year']].head()
```

```python
df['Registration_Year'] = df['Registration_Date'].dt.year
df[['Registration_Date','Registration_Year']].head()
```

```python
df['Registration_Year'] = df['Registration_Date'].dt.month
```

```python
df['Average_Order_Value'] = (
    df['Total_Spent']/df['Total_Orders']
)
```

```python
df[['Total_Orders','Total_Spent','Average_Order_Value']].head()
```

```python
def spending_category(x):
  if x < 10000:
    return 'Low'
  elif x < 50000:
    return 'Medium'
  else:
    return 'High'

df['Spending_Category'] = df['Total_Spent'].apply(spending_category)

df[['Total_Spent', 'Spending_Category']].head()
```

Question 1

How many customers are there in each gender category

```python
df['Gender'].value_counts()
```

```python
plt.figure(figsize = (16,5))
df['Gender'].value_counts().plot(
    kind='bar',
    color='#9850b2'
)
plt.tight_layout()
plt.title('Bar Chart : Customer Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.grid()
plt.savefig('GenderDistribution.png')
```

Question 2

What is the age distribution of customer

```python
plt.figure(figsize=(16,5))

df['Age'].plot(
    kind='hist',bins=10,
    color='#1c3ca9'
)

plt.tight_layout()
plt.title("Histogram: Age Distribution of Customer")
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.grid()
plt.savefig('AgeDistribution.png')
plt.show()
```

Question 3

How does customer spending vary?

```python
plt.figure(figsize=(16,5))

df['Total_Spent'].plot(
    kind='hist',bins=10,color='#a30096'
)
plt.tight_layout()
plt.title("Histogram: Distribution of Total Customer Spending")
plt.xlabel('Total Spent')
plt.ylabel('Number of Customer')
plt.grid()
plt.savefig("TotalSpentDistribution.png")
plt.show()
```

Question 4

Do customers with more orders spend more?


```python
plt.figure(figsize=(16,5))

plt.scatter(
    df['Total_Orders'],
    df['Total_Spent'],
    color='#862e68'
)

plt.tight_layout()
plt.title("Scatter Plot : Total Orders vs Total Spent")
plt.xlabel('Total Orders')

plt.ylabel("Total Spent")
plt.grid()
plt.savefig('OrderVsSpent.png')
plt.show()
```

Question 5

Which customer tier has the highest average spending?

```python
plt.figure(figsize = (16,5))

df.groupby('Customer_Tier')['Total_Spent'].mean().plot(
    kind='bar',
    color='#594e90'
)

plt.tight_layout()
plt.title('Bar Chart : Average Spending by Customer Tier')
plt.xlabel('Customer Tier')
plt.ylabel("Average Total Spent")
plt.xticks(rotation=360)
plt.grid()
plt.show()
```

Question 6

What is the percentage distribution of customers across Customer Tiers?

```python
plt.figure(figsize = (16,5))

df['Customer_Tier'].value_counts().plot(
    kind='pie',
    autopct="%1.1f%%"
)
plt.tight_layout()
plt.title('Pie Chart : Customer Distribution by Customer Tier')
plt.savefig('CustomerTierDistribution.png')
```

Question 7
How are customers distributed across Spending Categories?

```python
plt.figure(figsize = (16,5))

df['Spending_Category'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.tight_layout()

plt.title('Pie Chart : Customer Distribution by Spending Category')

plt.ylabel('')

plt.savefig('SpendingCategoryDistribution.png')
```

Question 8

How many customers registered in each year?

```python
registration_year = df['Registration_Year'].value_counts().sort_index()

plt.figure(figsize = (16,5))

registration_year.plot(
    kind='line',
    color='red',
    marker='o'
)

plt.tight_layout()

plt.title('Line Chart : Year Wise Customer Registration')

plt.xlabel('Registration Year')

plt.ylabel('Number of Customers')

plt.grid()

plt.savefig('YearWiseCustomerRegistration.png')
```

Question 9:

Which customer tier places the highest average number of orders?

```python
average_orders = df.groupby('Customer_Tier')['Total_Orders'].mean()

plt.figure(figsize = (16,5))

average_orders.plot(
    kind='bar',
    color='#0089ff'
)

plt.tight_layout()

plt.title('Bar Chart : Average Orders by Customer Tier')

plt.xlabel('Customer Tier')

plt.ylabel('Average Number of Orders')

plt.grid()

plt.savefig('AverageOrdersByCustomerTier.png')
```

Question 10

Does customer age have any relationship with total spending?



```python
plt.figure(figsize = (16,5))

plt.scatter(
    df['Age'],
    df['Total_Spent'],
    color = 'blueviolet'
)

plt.tight_layout()
plt.title('Scatter Plot : Age vs Total Spending')
plt.xlabel('Age')
plt.ylabel('Total Spent')
plt.grid()
plt.savefig('AgeVsTotalSpent.png')
```

Question 11
Do different customer tiers have different spending distributions?

```python
plt.figure(figsize = (16,5))

df.boxplot(
    column='Total_Spent',
    by='Customer_Tier'
)

plt.tight_layout()

plt.title('Box Plot : Spending Distribution by Customer Tier')

plt.xlabel('Customer Tier')

plt.ylabel('Total Spent')

plt.grid()

plt.savefig('SpendingByCustomerTier.png')
```
